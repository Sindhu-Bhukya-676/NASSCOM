```python
import sympy as sp


def g(x):
    return x**3 + 2 * x


# 1. Numerical derivative at x = 2 (use h = 1e-6)
x_val = 2.0
h = 1e-6

# Using Central Difference Formula: (g(x + h) - g(x - h)) / (2 * h)
num_deriv = (g(x_val + h) - g(x_val - h)) / (2 * h)
print(f"1. Numerical Derivative at x = {int(x_val)}: {num_deriv}")

# 2. Symbolic derivative of x**3 + 2*x
x_sym = sp.Symbol("x")
g_sym = x_sym**3 + 2 * x_sym
dg_dx_sym = sp.diff(g_sym, x_sym)
print(f"2. Symbolic Derivative: {dg_dx_sym}")

# 3. Evaluate the symbolic derivative at x = 2 and compare
sym_deriv_val = float(dg_dx_sym.subs(x_sym, x_val))
print(f"3. Symbolic Derivative evaluated at x = {int(x_val)}: {sym_deriv_val}")

diff = abs(num_deriv - sym_deriv_val)
print(f"   Absolute Difference: {diff:.10e}")
```

```python
f2 = x**2 + 3*x*y + y**2

# A partial derivative differentiates ONE variable, holding others fixed
print('df/dx =', sp.diff(f2, x))     # 2*x + 3*y
print('df/dy =', sp.diff(f2, y))     # 3*x + 2*y
```

```python
# The gradient stacks every partial derivative into one vector
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)

# Evaluate the gradient at the point (x=1, y=2)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)   # points in the steepest-ascent direction
```

```python
import sympy as sp

# Define symbolic variables
x, y = sp.symbols("x y")

# Function definition
h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print("1. Partial derivatives:")
print(f"   dh/dx = {dh_dx}")
print(f"   dh/dy = {dh_dy}")

# 2. Assemble the gradient list
grad = [dh_dx, dh_dy]
print(f"\n2. Gradient vector: {grad}")

# 3. Evaluate at (x=2, y=0)
grad_eval = [expr.subs({x: 2, y: 0}) for expr in grad]
print(f"\n3. Gradient evaluated at (x=2, y=0): {grad_eval}")
```

```python
import sympy as sp

x = sp.Symbol("x")
y = sp.exp(x**2 + 1)

# 1. By hand (as a SymPy expression using Chain Rule: inner' * outer')
inner_prime = 2 * x
outer_prime = sp.exp(x**2 + 1)
dy_dx_hand = inner_prime * outer_prime

print("1. Derivative by hand:      ", dy_dx_hand)

# 2. With sp.diff
dy_dx_auto = sp.diff(y, x)

print("2. Derivative via sp.diff:  ", dy_dx_auto)

# 3. Confirm they match
are_equal = dy_dx_hand.equals(dy_dx_auto)

print("\n3. Do the expressions match?", are_equal)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1                 # linear layer 1
h  = np.maximum(0, z1)      # ReLU activation
y_hat = h @ W2              # linear layer 2 (prediction)
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1               # step downhill
W2 -= lr * dW2

# Recompute the loss after the update
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1  # input -> hidden
Wb = np.random.randn(8, 1) * 0.1  # hidden -> output

# 1. Forward pass: z1, h = ReLU(z1), y_hat, loss
z1 = Xb @ Wa
h = np.maximum(0, z1)  # ReLU activation
y_hat = h @ Wb
loss_before = np.mean((y_hat - Yb) ** 2)

# 2. Backward pass: dy, dWb, dh, dz1, dWa
N = Xb.shape[0]
dy = 2 * (y_hat - Yb) / N  # Derivative of MSE loss w.r.t. y_hat
dWb = h.T @ dy  # Gradient w.r.t. Wb
dh = dy @ Wb.T  # Gradient w.r.t. hidden layer h
dz1 = dh * (z1 > 0)  # Derivative through ReLU
dWa = Xb.T @ dz1  # Gradient w.r.t. Wa

# 3. One step with lr = 0.05; print loss before and after
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb

# Forward pass after update to check new loss
z1_new = Xb @ Wa
h_new = np.maximum(0, z1_new)
y_hat_new = h_new @ Wb
loss_after = np.mean((y_hat_new - Yb) ** 2)

print(f"Loss before update: {loss_before:.6f}")
print(f"Loss after update:  {loss_after:.6f}")
```

```python
xv = 0.0          # starting point
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)        # the derivative
    xv = xv - lr * grad        # step against the gradient
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```
