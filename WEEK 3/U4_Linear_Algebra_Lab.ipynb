```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)


S = np.array([[1.5, 0.0],
              [0.0, 0.5]])


t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    # repeat the first point at the end so the polygon closes
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
  import matplotlib.pyplot as plt
import numpy as np

tri = np.array([[0, 1, 0.5], [0, 0, 1.0]])


S = np.array([[2.0, 0.0], [0.0, 0.5]])

# 2. Rotation matrix (45 degrees)
theta = np.radians(45)
R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])

# 3. Apply both and plot original vs transformed
# Multiply matrices: R @ S @ tri applies scaling first, then rotation
tri_transformed = R @ S @ tri

# Append the first vertex to the end of each array to close the triangle loop
tri_orig_plot = np.hstack([tri, tri[:, :1]])
tri_trans_plot = np.hstack([tri_transformed, tri_transformed[:, :1]])

# Plotting
plt.figure(figsize=(7, 7))
plt.plot(
    tri_orig_plot[0],
    tri_orig_plot[1],
    "b-o",
    label="Original",
    linewidth=2,
)
plt.plot(
    tri_trans_plot[0],
    tri_trans_plot[1],
    "r-o",
    label="Transformed (Scaled + Rotated)",
    linewidth=2,
)

# Reference axes and grid setup
plt.axhline(0, color="gray", linestyle="--", alpha=0.7)
plt.axvline(0, color="gray", linestyle="--", alpha=0.7)
plt.grid(True, linestyle=":", alpha=0.6)
plt.axis("equal")
plt.legend()
plt.title("2D Transformation: Scaling & Rotation")
plt.show()
```

```python

A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np

C = np.array([[4.0, 1.0], [2.0, 3.0]])

# 1. eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

# 2. print eigenvalues
print("Eigenvalues:", eigenvalues)

# 3. verify C v = lambda v for the first eigenvector
lambda_1 = eigenvalues[0]
v_1 = eigenvectors[:, 0]  # First column corresponds to the first eigenvalue

lhs = C @ v_1
rhs = lambda_1 * v_1

print("\nVerification for the first eigenvector:")
print("C @ v_1:       ", lhs)
print("lambda_1 * v_1:", rhs)
print("Are equal?     ", np.allclose(lhs, rhs))
```

```python
A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)                    # [2, 3, -1]
print('Rank of A :', la.matrix_rank(A))     # 3 -> full rank
print('Full rank -> a unique solution exists')
```

```python
import numpy as np

A2 = np.array([[3.0, 2.0], [1.0, 4.0]])
b2 = np.array([7.0, 9.0])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])

# 1. Solve A2 x = b2
x = np.linalg.solve(A2, b2)
print("1. Solution x:", x)

# 2. Rank of A2
rank = np.linalg.matrix_rank(A2)
print("2. Rank of A2:", rank)


# 3. cosine(cat, dog) and cosine(cat, car); which is more similar?
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))


cos_cat_dog = cosine_similarity(cat, dog)
cos_cat_car = cosine_similarity(cat, car)

print(f"3. Cosine(cat, dog): {cos_cat_dog:.4f}")
print(f"   Cosine(cat, car): {cos_cat_car:.4f}")

if cos_cat_dog > cos_cat_car:
    print("   -> 'cat' and 'dog' are more similar.")
else:
    print("   -> 'cat' and 'car' are more similar.")
```
